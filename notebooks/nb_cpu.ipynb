```python
import os
import sys
from pathlib import Path
from dotenv import load_dotenv

# 1. Xác định thư mục gốc của dự án (project root)
# Nếu đang đứng ở thư mục 'notebooks', lùi ra 1 cấp về thư mục cha
REPO_DIR = Path.cwd().resolve()
if REPO_DIR.name == "notebooks":
    REPO_DIR = REPO_DIR.parent

# 2. Chuyển thư mục làm việc về gốc dự án (để các đường dẫn tương đối data/, configs/ luôn đúng)
os.chdir(REPO_DIR)

# 3. Thêm thư mục 'src' vào sys.path để import các module trong dự án
src_dir = REPO_DIR / "src"
if src_dir.exists() and str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

# 4. Tải biến môi trường từ file .env (nếu có)
load_dotenv()
print(f"💻 Thư mục làm việc: {REPO_DIR}")
```
