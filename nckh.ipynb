### Đồng bộ file Notebook hiện tại lên GitHub
Các lệnh dưới đây sẽ tìm kiếm các file `.ipynb` hoạt động trong Google Drive của bạn, sao chép chúng vào thư mục dự án gốc `/content/drive/MyDrive/NCKH_PanDerm` và tự động đẩy lên GitHub.

```python
import os
import shutil
import glob

# Đường dẫn đích trong thư mục dự án của bạn
DEST_DIR = "/content/drive/MyDrive/NCKH_PanDerm"

# Tìm kiếm tất cả các file notebook (.ipynb) trong thư mục mặc định "Colab Notebooks" trên Drive của bạn
colab_notebooks_path = "/content/drive/MyDrive/Colab Notebooks/*.ipynb"
notebooks = glob.glob(colab_notebooks_path)

print("--- ĐANG TÌM KIẾM NOTEBOOK ĐỂ SAO CHÉP ---")
if notebooks:
    for nb in notebooks:
        # Lấy tên file gốc
        nb_name = os.path.basename(nb)
        # Bạn có thể lọc hoặc đổi tên file nếu muốn tại đây
        destination_path = os.path.join(DEST_DIR, nb_name)
        shutil.copy(nb, destination_path)
        print(f"Đã sao chép thành công notebook: {nb_name} vào thư mục dự án.")
else:
    print("Không tìm thấy file notebook nào trong thư mục mặc định 'Colab Notebooks'.")
    print("Nếu bạn đặt tên file notebook này khác, hãy lưu ý bấm 'File -> Save' (Ctrl+S) trên Colab trước để Drive cập nhật tệp tin mới nhất.")
```
